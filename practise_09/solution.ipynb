```python
import pandas as pd

FILE = "Sample - Superstore Sales.xls"

orders = pd.read_excel(FILE, sheet_name="Orders")
returns = pd.read_excel(FILE, sheet_name="Returns")
users = pd.read_excel(FILE, sheet_name="Users")
```

```python
#Завдання 1. Завантаження і перший огляд
print(pd.__version__) #версія pandas
print(orders.shape, returns.shape, users.shape) #розмір таблиць
pd.set_option('display.max_columns', None) #налаштування, щоб бачити усі поля
print(orders.head(5)) # перші 5 рядків
print(orders.tail(5)) #останні 5 рядків
print(orders.shape) #розмір orders
print(orders.dtypes) #типи даних полів
print(orders.memory_usage(deep=True).sum()) #скільки мегабайтів пам’яті займає таблиця
print(orders.memory_usage(deep=True).idxmax()) #назва колонки, яка займає найбільше пам'яті
```

```python
#Завдання 2. Описова статистика
print(orders.describe(include='number')) #описова статистика для числових полів
print(orders.describe(exclude='number')) #описова статистика для нечислових полів
print(len(orders['Product Category'].unique()), len(orders['Product Sub-Category'].unique()), len(orders['Region'].unique())) # кількість унікальних значень у колонках Product Category, Product Sub-Category і Region
print(round(orders['Ship Mode'].value_counts(normalize=True) * 100, 2)) #частка кожного способу доставки Ship Mode
```

```python
#Завдання 3. Вибірка й індексація
print(orders[['Order ID', 'Sales', 'Profit']].equals(orders.loc[:, ['Order ID', 'Sales', 'Profit']])) #отримайте підтаблицю лише з колонок Order ID, Sales і Profit двома різними способами і переконайтеся, що результати однакові
print(orders.iloc[100:111]) #рядки з позиціями від 100 до 110 включно
print(orders.loc[0:20, 'Profit':'Product Category']) #перші 20 рядків і колонки від Profit до Product Category включно
orders_copy = orders.copy() #робимо копію дф
orders_copy.index = orders_copy['Order ID']#змінюємо індекси на Order ID
print(len(orders_copy.loc[24132]))#кі-ть рядків замовлення 24132
orders_copy.reset_index(drop=True) #скидаємо індекси
print(orders.shape == orders_copy.shape) #розміри таблиці не змінилися
```

```python
#Завдання 4. Фільтрація
print(len(orders[orders['Profit'] < 0])) #усі збиткові записи, де Profit менший за нуль, і виведіть їхню кількість
print(orders[(orders['Product Category'] == 'Technology') & (orders['Sales'] > 100)]) #замовлення категорії Technology із Sales понад 1000
print(orders[orders['Region'].isin(['West', 'Atlantic'])]) #замовлення з регіонів West або Atlantic, не повторюючи умову двічі
print(orders[orders['Discount'].between(0.05, 0.1)]) #замовлення зі знижкою в діапазоні від 0.05 до 0.10 включно
print(orders[orders['Province'].str.startswith('S')]['Province'].unique()) #замовлення, де назва провінції у Province починається на літеру S, і виведіть перелік унікальних провінцій у результаті
```

```python
#Завдання 5. Пропуски, дублікати та типи даних
na = orders.isna().sum() #кількість порожніх значень у кожній колонці
print(na[na > 0].index[0]) #колонки, де пропуски справді є
orders_copy_2 = orders.copy() #копія ДФ
orders_copy_2['Product Base Margin'].fillna(orders_copy_2['Product Base Margin'].mode()[0], inplace=True) # заповнюємо пропуски найчастішим значенням
print(orders_copy_2['Product Base Margin'].mode()[0]) #виводимо моду по Product Base Margin
orders_copy_3 = orders.dropna(subset='Product Base Margin') #копія orders з видаленими пустими значеннями
print(len(orders), len(orders_copy_3)) #ількість рядків до і після видалення
print(orders.duplicated().sum()) #кількість повністю однакових рядків
print(orders.duplicated(subset=['Order ID', 'Product Name']).sum()) #кількість рядків, що повторюються за парою Order ID + Product Name
order_id_copy = orders['Order ID'].astype('str') #Order ID на текстовий тип
discount_copy = orders['Discount'].astype('float32') #Discount на float32
print(orders['Order ID'].memory_usage(deep=True), order_id_copy.memory_usage(deep=True)) #виводимо обсяг пам’яті до і після перетворення
print(orders['Discount'].memory_usage(deep=True), discount_copy.memory_usage(deep=True)) #виводимо обсяг пам’яті до і після перетворення
```

```python
#Завдання 6. Робота з текстом
orders_copy_4 = orders.copy() #копія ДФ
orders_copy_4['First Name'] = orders_copy_4['Customer Name'].str.split(' ', n=1, expand=True).iloc[:,0] #ділимо Customer Name та беремо 1 поле
orders_copy_4['Last Name'] = orders_copy_4['Customer Name'].str.split(' ', n=1, expand=True).iloc[:,1] #ділимо Customer Name та беремо 2 поле
print(orders_copy_4['Last Name'].isna().sum()) #перевіряємо ну пусті значення
orders_copy_4['Name Length'] = orders_copy_4['Product Name'].str.len() #довжина назви товару в символах
print(orders_copy_4['Name Length'].min(), orders_copy_4['Name Length'].max()) #найкоротша й найдовша довжина
print(orders_copy_4['Product Name'].str.contains('Chair', case=False).sum()) #скільки рядків містять у назві товару слово Chair
orders_copy_4['Province'] = orders_copy_4['Province'].str.replace('Saskachewan', 'Saskatchewan') #Saskachewan міняємо на Saskatchewan
orders_copy_4['Product Sub-Category'] = orders_copy_4['Product Sub-Category'].str.replace('&', 'and', regex=False) #символ & на слово and
print(orders_copy_4['Province'].unique()) # унікальні значення Province
print(orders_copy_4['Product Sub-Category'].unique()) #унікальні значення Product Sub-Category
```

```python
#Завдання 7. Дати й час
print(orders['Order Date'].dtype, orders['Ship Date'].dtype)
orders_copy_5 = orders.copy() #копія ДФ
orders_copy_5['Order Year'] = orders_copy_5['Order Date'].dt.year #визначаємо поле Order Year
orders_copy_5['Order Month'] = orders_copy_5['Order Date'].dt.month #визначаємо поле Order Month
orders_copy_5['Order Weekday'] = orders_copy_5['Order Date'].dt.day_name #визначаємо поле Order Weekday
orders_copy_5['Order Quarter'] = orders_copy_5['Order Date'].dt.quarter #визначаємо поле Order Quarter
print(orders_copy_5.groupby('Order Year')['Order ID'].count()) #кількість замовлень за кожним роком
orders_copy_5['Ship Delay'] = orders_copy_5['Ship Date'] - orders_copy_5['Order Date'] #
orders_copy_5['Ship Delay'] = orders_copy_5['Ship Delay'].dt.days #кільки днів минуло між замовленням і відвантаженням
print(orders_copy_5['Ship Delay'].min(), orders_copy_5['Ship Delay'].max(), orders_copy_5['Ship Delay'].mean()) # мінімум, максимум і середнє
orders_copy_rsm = orders_copy_5[['Sales', 'Order Date']].set_index('Order Date') #новий дф з індексом Order Date
orders_copy_rsm = orders_copy_rsm.resample('ME').sum() #помісячний ряд сумарних продажів: індекс — кінець місяця
print(len(orders_copy_rsm)) #кількість точок у ряді
orders_copy_rsd = orders_copy_5[['Ship Delay', 'Order Date']].set_index('Order Date') #новий дф з індексом Order Date
orders_copy_rsd.resample('D').mean() #середня затримка відвантаження за кожним днем тижня замовлення
```

```python
#Завдання 8. Похідні колонки
orders_copy_5.assign(
    Profit_Margin = orders_copy_5['Profit'] / orders_copy_5['Sales'],
    Revenue_Per_Unit = orders_copy_5['Sales'] / orders_copy_5['Order Quantity'],
    Is_Profitable =  orders_copy_5['Profit'] > 0
) #додаємо колонки одним виразом
orders_copy_5['Discount Level'] = orders_copy_5['Discount'].apply(lambda x: 'No' if x == 0  else 'Low' if x < 0.05 else 'High') # додаємо Discount Level через apply
orders_copy_5['Discount Level_'] = pd.cut(orders_copy_5['Discount'], bins=[-1, 0, 0.04, 1], labels=['No', 'Low', 'High']) # додаємо Discount Level_ через pd.cut
print((orders_copy_5['Discount Level'] != orders_copy_5['Discount Level_']).sum()) #перевіряємо на рівність значень
orders_copy_5.drop('Discount Level_', inplace=True, axis=1) #видаляємо дублюючу колонку Discount Level_
mappings = {'West':'W', 'Atlantic':'A', 'Ontario':'ON', 'Quebec':'QC', 'Prarie':'PRA', 'Nunavut':'NU', 'Northwest Territories': 'NWT', 'Yukon':'YT'} #словник для map
orders_copy_5['Region Code'] = orders_copy_5['Region'].map(mappings) # map по словнику
print(orders_copy_5['Region Code'].isna().sum()) #перевірка та пусті значення
orders_copy_5['Delivery Speed'] = orders_copy_5['Ship Delay'].apply(lambda x: 'швидке' if x <= 2 else 'звичайне' if x <= 5 else 'повільне') #Delivery Speed через apply
print(orders_copy_5.groupby(['Delivery Speed'])['Delivery Speed'].count()) #кількість записів у кожній категорії
```

```python
#Завдання 9. Сортування та ранжування
print(orders['Sales'].sort_values(ascending=False).iloc[0]) #сортування за Sales за спаданням, вибираємо найбільше значення продажів
orders.sort_values(['Region', 'Profit'], ascending=[True, False]) #сортування за двома ключами одночасно: Region за зростанням, Profit за спаданням
print(orders.nlargest(10, ['Profit'])) #10 найприбутковіших записів
print(orders.nsmallest(10, ['Profit'])) #10 найзбитковіших записів
orders_copy_5['Sales Rank'] = orders_copy_5.groupby('Region')['Sales'].rank(method='min', ascending=False) #поле Sales Rank
print(orders_copy_5[(orders_copy_5['Region'] == 'West') & (orders_copy_5['Sales Rank'] == 1)]) #запис із рангом 1 для регіону West
```

```python
#Завдання 10. Групування та агрегація
print(orders.groupby('Product Category').agg({'Sales': 'sum', 'Profit': 'sum'})) #сумарні Sales і Profit за кожною Product Category
print(orders.groupby('Region').agg(total_sales=('Sales', 'sum'), avg_discount=('Discount', 'mean'), orders_count=('Order ID', 'nunique'), max_profit=('Profit', 'max'))) #чотири показники через agg
orders_prcat = orders.groupby(['Product Category', 'Product Sub-Category'])['Profit'].sum() #сумарний прибуток за кожною парою Product Category + Product Sub-Categor
print(orders_prcat) #виводимо orders_prcat
orders_copy_5['Category Avg Sales']  =  orders_copy_5.groupby('Product Category')['Sales'].transform('mean') #поле Category Avg Sales
orders_copy_5['Sales Deviation'] = orders_copy_5['Sales'] - orders_copy_5['Category Avg Sales'] #поле Sales Deviation'
pvt = pd.pivot_table(orders, values='Profit', index='Product Sub-Category', columns='Product Category', aggfunc='sum') #робимо pivot
print(pvt.idxmax()) # підкатегорії з найбільшим сумарним прибутком
```

```python
#Завдання 11. Зведені таблиці та зміна форми
print(pd.pivot_table(orders, values='Sales', index='Region', columns='Product Category', aggfunc='sum', margins=True)) #зведена таблицю сумарних продажів
print(pd.crosstab(orders['Customer Segment'], orders['Ship Mode'])) #таблиця спряженості кількості записів
#по 11.3 не зрозумів, що треба зробити
```

```python
#Завдання 12. Об’єднання таблиць
orders_mrg_returns = pd.merge(orders, returns, on='Order ID', how='left') #робимо merge
orders_mrg_returns['Status'] = orders_mrg_returns['Status'].fillna('No') #застосовуємо fillna
print(len(orders_mrg_returns)) #к-ть рядків
print(orders_mrg_returns['Status'].value_counts()) #розподіл значень поля Status
print(users['Region'].unique(), orders['Region'].unique()) #значення поля Region у users
print(users['Region'].value_counts()) #розподіл значень поля Region у users
orders_mrg_users_1 = pd.merge(orders, users, on='Region', how='left') #об'єднання до видалення повторів
users_copy = users.drop_duplicates(subset='Region', keep='last') #копія users без повторів
orders_mrg_users_2 = pd.merge(orders, users_copy, on='Region', how='left') #об'єднання після видалення повторів
print(len(orders_mrg_users_1), len(orders_mrg_users_2)) #кількість рядків результату
print(orders_mrg_users_1['Manager'].notna().sum(), orders_mrg_users_2['Manager'].notna().sum()) # кількість записів, які отримали менеджера
df_1= pd.DataFrame({'id': [1, 2, 3], 'name': ['a', 'b', 'c']}) #перший дф
df_2= pd.DataFrame({'id': [1, 4, 5], 'name_d': ['aa', 'bb', 'cc']}) #другий дф
mrg_inner = pd.merge(df_1, df_2, on='id', how='inner') #merge inner
mrg_left = pd.merge(df_1, df_2, on='id', how='left') #merge left
mrg_right = pd.merge(df_1, df_2, on='id', how='right') #merge right
mrg_outer = pd.merge(df_1, df_2, on='id', how='outer') #merge outer
print('inner', len(mrg_inner)) #кількість рядків inner
print('left', len(mrg_left)) #кількість рядків left
print('right', len(mrg_right)) #кількість рядків right
print('outer', len(mrg_outer)) #кількість рядків outer
A = orders_copy_5[orders_copy_5['Order Year'] <= 2010] #беремо копію, де вже є поле Order Year
B = orders_copy_5[orders_copy_5['Order Year'] > 2010] #беремо копію, де вже є поле Order Year
C = pd.concat([A, B], axis=0) #об'єднання за стороками
print(len(orders_copy_5), len(C)) #перевіряємо на к-ть рядків
```

```python
#Завдання 13. Бінінг, категорії та накопичувальні обчислення
maxx = orders['Sales'].max() #максимальні продажі для межі останнього інтервалу
sales_cut = pd.cut(orders['Sales'], bins=[0, 500, 2000, 10000, maxx], labels=['Low', 'Medium', 'High', 'Very High']) #ділимо дф
print(sales_cut.value_counts()) #розподіл по інтервалам
sales_cut_q = pd.cut(orders['Sales'], bins=4, labels=['Low', 'Medium', 'High', 'Very High']) #ділимо на чотири рівні за шириною інтервали
print(sales_cut_q.value_counts()) # розподіл по інтервалам
orders_copy_5['Profit_Q'] = pd.qcut(orders_copy_5['Profit'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4']) #ділимо Profit на 4 рівні інтервали
orders_copy_5[['Profit_Q', 'Sales']].groupby('Profit_Q')['Sales'].mean() #середні продажі в кожному інтервалі
orders_copy_5['Product Category'] = orders_copy_5['Product Category'].astype('category') #змінюємо тип даних на category
orders_copy_5['Region'] = orders_copy_5['Region'].astype('category') #змінюємо тип даних на category
print(orders['Product Category'].memory_usage(), orders_copy_5['Product Category'].memory_usage(),
      orders_copy_5['Product Category'].memory_usage() / orders['Product Category'].memory_usage()) #виводимо обсяг пам’яті Product Category до і після перетворення та у скільки разів він змінився
print(orders['Region'].memory_usage(), orders_copy_5['Region'].memory_usage(), orders_copy_5['Region'].memory_usage() / orders['Region'].memory_usage()) #виводимо обсяг пам’яті Region до і після перетворення та у скільки разів він змінився
orders_copy_5.sort_values('Order Date', inplace=True) #сортуємо на копії
orders_copy_5['Sales_Cumsum'] = orders_copy_5['Sales'].cumsum() #додаємо наростаючу суму продажів
print(round((orders_copy_5['Sales_Cumsum'].iloc[orders_copy_5.shape[0] - 1]), 2) == round(orders_copy_5['Sales'].sum(), 2)) #округлюємо до 2 знаків, внакше рівності не буде
roll_mean = orders_copy_rsm.rolling(3).mean() #ковзне середнє з вікном у три місяці,
exp_mean = orders_copy_rsm.expanding(1).mean() #наростаюче середнє від початку періоду
pct_change = orders_copy_rsm.pct_change() #відсоткова зміна продажів
diff = orders_copy_rsm.diff() #відсоткова зміна продажів
print(roll_mean.isna().sum(), exp_mean.isna().sum(), pct_change.isna().sum(), diff.isna().sum()) #кількість порожніх значень на початку кожного з цих рядів
```
