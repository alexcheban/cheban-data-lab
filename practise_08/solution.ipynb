```python
import pandas as pd
import numpy as np

data = {
    "source":  ["web", "web", "store", "app", "app", "web", "store", "web", "app", "app"],
    "amount":  [250.0, 1300.0, 90.5, np.nan, 640.0, 180.0, 415.0, np.nan, 2100.0, 320.0],
    "items":   [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    "express": ["yes", "yes", "no", "yes", "no", "no", "no", "yes", "no", "no"],
}

labels = ["ord_1", "ord_2", "ord_3", "ord_4", "ord_5",
          "ord_6", "ord_7", "ord_8", "ord_9", "ord_10"]
```

```python
#Завдання 1. Створення таблиці
orders = pd.DataFrame(data) #створюємо дф
print(orders) # виводимо дф
orders.index = labels #змінюємо індекси на значення зі списку
print(orders)
print(orders.shape) #виводимо розмір дф
print(orders.dtypes) #виводимо типи полів
```

```python
#Завдання 2. Перший погляд на дані
print(orders.head(3)) #виводимо перші 3 рядки
print(orders.tail(2)) #виводимо останні 2 рядки
print(orders['amount'].count()) #виводимо к-ть непустих значень поля amount
print(orders.isna().sum()) #виводимо к-ть пустих значень по усім полям
print(orders.describe()) #виводимо статистики числових полів
```

```python
#Завдання 3. Вибірка колонок
print(orders['source']) #виводимо поле source
print(orders[['source', 'amount']]) #виводимо поля 'source', 'amount'
print(orders.columns)  #виводимо назви полів
print(orders['source'].unique()) #виводимо унікальні значення поля source
```

```python
#Завдання 4. Рядки за позицією
print(orders.iloc[0]) #виводимо найперший рядок таблиці
print(orders.iloc[[0, 2, 3]]) #виводимо рядки з позиціями 0, 2 і 3
print(orders.iloc[[0, 2, 3], [0, 1]]) #виводимо рядки з позиціями 0, 2, 3, але тільки колонки source і amount
print(orders.iloc[1:4]) #виводимо рядки з позиціями від 1 до 3 включно, використавши зріз
```

```python
#Завдання 5. Рядки за міткою
print(orders.loc['ord_5', 'source']) #виводимо значення, яке стоїть у колонці source у рядку з міткою ord_5
print(orders.loc[['ord_1', 'ord_3', 'ord_6']]) #виводимо рядки з мітками ord_1, ord_3 і ord_6
print(orders.loc[['ord_1', 'ord_3', 'ord_6'], ['source', 'items']]) #виводимо рядки з тими самими мітками, але тільки колонки source і items
```

```python
#Завдання 6. Фільтрація за однією умовою
print(orders[orders['amount'] > 500]) #виводимо  замовлення, де amount більша за 500
print(orders[orders['source'] == 'web']) #виводимо замовлення з каналу web
print(orders[orders['express'] == 'yes']) #виводимо замовлення зі швидкою доставкою, тобто де express дорівнює yes
print(orders[orders['source'] != 'app']) #виводимо замовлення, які не з каналу app
print(orders[orders['source'] != 'app'].shape[0]) #виводимо кількість рядків у результаті пункту 6.4.
```

```python
#Завдання 7. Фільтрація за кількома умовами
print(orders[(orders['source'] == 'app') & (orders['items'] > 1)]) #виводимо замовлення з каналу app, у яких items більше за 1
print(orders[(orders['express'] == 'no') & (orders['amount'] >= 400)]) #виводимо  замовлення без швидкої доставки, у яких amount більша або дорівнює 400
print(orders[(orders['amount'] < 300) | (orders['amount'].isna())]) #виводимо замовлення, де amount менша за 300 або відсутня
print(orders[orders['source'].isin(['web', 'store'])]) #виводимо замовлення, канал яких — web або store, без повторення умови двічі
print(orders[orders['items'].between(2, 3)]) #виводимо замовлення, де items у діапазоні від 2 до 3 включно, скориставшись спеціальним методом для діапазонів
print(orders[orders['amount'].between(200, 700)]) #виводимо замовлення, де amount у діапазоні від 200 до 700 включно
print(orders[orders['items'] == 1].loc[:, ['source', 'items']]) #виводимо тільки колонки source і items для замовлень, у яких items дорівнює 1
```

```python
#Завдання 8. Пропущені значення
print(orders[orders['amount'].notna()]) #виводимо замовлення, у яких amount заповнена
print(orders[orders['amount'].isna()]) #виводимо замовлення, у яких amount відсутня
orders_copy = orders.fillna(orders['amount'].median()) #копія таблиці, міняємо Nan на медіану
print(orders['amount'].median()) #виводимо медіану
print(orders_copy['amount'].isna().sum()) #перевіряємо, що в новій таблиці немає Nan
orders_copy_1 = orders.dropna(subset='amount') #копія таблиці, з якої видалені рядки з порожнім amount
print(orders_copy_1.shape) #виводимо розміри таблиці
print(orders)
```

```python
#Завдання 9. Зміна таблиці
work = orders.copy() #робимо копію
work['amount'] = work['amount'] + 50 # збільшуємо в ній усі значення amount на 50
print(orders.loc['ord_1', 'amount'], work.loc['ord_1', 'amount']) # виводимо значення amount для ord_1 в обох таблицях
work['avg_item'] = round(work['amount'] / work['items'], 2) # додаємо в work нову колонку avg_item
work.rename(columns={'source':'channel'}, inplace=True) #Перейменуйте в work колонку source на channel
work.drop(columns=['express'], inplace=True) #видаляємо поле express
print(work.columns) #виводимо список полів
```

```python
#Завдання 10. Сортування та підсумки
print(orders.sort_values('amount', ascending=0)) #виводимо результат сортування
orders.sort_values(['source', 'amount']) #сортуємо таблицю спершу за source, а всередині кожного каналу — за amount за зростанням
print(orders['amount'].sum()) #виводимо суму всіх значень колонки amount
print(round(orders['amount'].mean(), 2)) #виводимо середнє значення колонки amount, округлене до двох знаків
print(orders[orders['amount'] == orders['amount'].max()]) #виводимо рядок або рядки з максимальним значенням amount
print(orders.idxmax()['amount']) #виводимо мітку рядка з максимальним amount
```
